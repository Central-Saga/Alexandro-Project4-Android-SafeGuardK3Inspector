# SafeGuard K3 Inspector – Exploratory Data Analysis (EDA)
### Analisis Distribusi Kelas & Dataset Kepatuhan APD Pekerja (K3)

Notebook ini mengeksplorasi dataset deteksi objek APD K3 yang digunakan untuk melatih model **YOLOv8**.

**6 Kelas Target K3:**
1. `Hard_hat` (Helm Keselamatan)
2. `Vest` (Rompi Reflektif / Safety Vest)
3. `Person` (Pekerja / Manusia)
4. `Mask` (Masker)
5. `Gloves` (Sarung Tangan)
6. `Safety_boots` (Sepatu Pelindung K3)

```python
import os
import glob
import yaml
from collections import Counter
import matplotlib.pyplot as plt

# Definisi kelas kanonikal K3
CLASSES = ['Gloves', 'Hard_hat', 'Mask', 'Person', 'Safety_boots', 'Vest']
print(f"Target Kelas K3 ({len(CLASSES)} kelas): {CLASSES}")
```

### 1. Struktur Dataset YOLOv8

```python
def check_dataset_structure(data_dir="dataset"):
    for split in ['train', 'valid', 'test']:
        img_path = os.path.join(data_dir, split, 'images')
        lbl_path = os.path.join(data_dir, split, 'labels')
        n_imgs = len(glob.glob(f"{img_path}/*.*g")) if os.path.exists(img_path) else 0
        n_lbls = len(glob.glob(f"{lbl_path}/*.txt")) if os.path.exists(lbl_path) else 0
        print(f"Split [{split.upper()}]: {n_imgs} gambar | {n_lbls} file label")

check_dataset_structure()
```

### 2. Distribusi Label Objek K3

```python
def analyze_class_frequencies(label_dir="dataset/train/labels"):
    class_counts = Counter()
    if not os.path.exists(label_dir):
        print(f"Direktori {label_dir} belum tersedia di lingkungan lokal (unduh dataset terlebih dahulu).")
        return
    for label_file in glob.glob(f"{label_dir}/*.txt"):
        with open(label_file, 'r') as f:
            for line in f:
                parts = line.strip().split()
                if parts:
                    class_counts[int(parts[0])] += 1
    print("Frekuensi kelas dalam subset pelatihan:")
    for cls_id, cnt in sorted(class_counts.items()):
        name = CLASSES[cls_id] if cls_id < len(CLASSES) else f"Unknown ({cls_id})"
        print(f" - {name}: {cnt} anotasi")

analyze_class_frequencies()
```
